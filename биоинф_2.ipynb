**Результат:** Топ-1 хит — **белок RHO (родопсин) сетчатки глаза** (opsin, rhodopsin)

Вывод: последовательность соответствует фрагменту родопсина — светочувствительного белка, участвующего в зрении.

### 2. Поиск ортолога у человека

**Выбранный метод:** `tblastn`

**Обоснование выбора:** tblastn сравнивает белковую последовательность с нуклеотидной базой данных, переводя её в 6 рамок трансляции. Это необходимо, потому что ортолог у человека может иметь другую нуклеотидную последовательность из-за вырожденности генетического кода, а также потому что некоторые участки генома могут быть не полностью аннотированы. blastp ищет только по уже аннотированным белкам и может пропустить неканонические варианты.

**База данных:** refseq_genomic (Homo sapiens)

**Результат:** Найден ортолог — **родопсин (RHO)**

### 3. Дефект зрения

Согласно базе OMIM, с белком RHO (родопсин) связан следующий дефект зрения:

> **Пигментный ретинит (Retinitis pigmentosa)** — наследственное заболевание, характеризующееся прогрессирующей дегенрацией фоторецепторов сетчатки, что приводит к снижению зрения в сумерках ("куриная слепота"), сужению полей зрения и в конечном итоге к слепоте. Мутации в гене RHO являются одной из наиболее частых причин аутосомно-доминантного пигментного ретинита.

cut -f2 luciferase_hits.txt | head -10 > top10_ids.txt

seqtk subseq transcriptome_photinus.fasta.transdecoder.pep top10_ids.txt > top10_luciferases.fasta
